```python
import pandas as pd

data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
]

columns = ['Sky', 'Temperature', 'Humidity', 'Wind', 'Water', 'Forecast', 'EnjoySport']

df = pd.DataFrame(data, columns=columns)

print(df)
```

```python
# Start with the first positive example
hypothesis = None

for i in range(len(df)):

    # Consider only positive examples
    if df.iloc[i]['EnjoySport'] == 'Yes':

        example = df.iloc[i][:-1].tolist()

        # First positive example becomes the initial hypothesis
        if hypothesis is None:
            hypothesis = example

        else:
            # Compare with current hypothesis
            for j in range(len(hypothesis)):
                if hypothesis[j] != example[j]:
                    hypothesis[j] = '?'

print("Final Hypothesis:")
print(hypothesis)
```

```python
attributes = columns[:-1]

print("Most Specific Hypothesis:")

for i in range(len(attributes)):
    print(attributes[i], "=", hypothesis[i])
```
