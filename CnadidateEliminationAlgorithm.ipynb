```python
import pandas as pd

data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
]

columns = ['Sky', 'Temperature', 'Humidity', 'Wind', 'Water', 'Forecast', 'EnjoySport']

df = pd.DataFrame(data, columns=columns)

print(df)
```

```python
S = ['0', '0', '0', '0', '0', '0']

G = [['?', '?', '?', '?', '?', '?']]

print("Initial S:", S)
print("Initial G:", G)
```

```python
for i in range(len(df)):

    example = df.iloc[i][:-1].tolist()
    label = df.iloc[i]['EnjoySport']

    if label == 'Yes':

        # Generalize S
        for j in range(len(S)):
            if S[j] == '0':
                S[j] = example[j]
            elif S[j] != example[j]:
                S[j] = '?'

    else:

        # Specialize G for negative example
        new_G = []

        for hypothesis in G:
            for j in range(len(example)):
                if hypothesis[j] == '?':
                    if S[j] != example[j]:
                        new_hypothesis = hypothesis.copy()
                        new_hypothesis[j] = S[j]
                        new_G.append(new_hypothesis)

        if new_G:
            G = new_G

print("Final Specific Hypothesis (S):")
print(S)

print("\nFinal General Hypothesis (G):")
for g in G:
    print(g)
```

```python
print("Specific Boundary (S):")
print(S)

print("\nGeneral Boundary (G):")
for g in G:
    print(g)
```
